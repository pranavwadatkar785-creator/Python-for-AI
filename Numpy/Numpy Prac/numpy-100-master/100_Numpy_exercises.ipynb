#### 35. How to compute ((A+B)*(-A/2)) in place (without copy)? (★★☆)

````python
A = np.array([10,20,30], dtype=float)
B = np.array([40,50,60], dtype=float)

np.add(A,B,out=B)
np.divide(A,-2,out=A)
np.multiply(B,A,out=A)
print(A)
````
